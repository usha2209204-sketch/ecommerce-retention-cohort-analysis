# E-commerce Retention and Cohort Analysis

This notebook explores a synthetic e-commerce dataset to answer the following questions:
- Which customer segments generate repeat revenue?
- Which cohorts retain customers best?
- Which products drive repeat purchases?
- How should retention campaigns be prioritized?


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

customers = pd.read_csv('../data/customers.csv')
orders = pd.read_csv('../data/orders.csv')
order_items = pd.read_csv('../data/order_items.csv')
products = pd.read_csv('../data/products.csv')

df = orders.merge(customers, on='customer_id', how='left')
df['order_date'] = pd.to_datetime(df['order_date'])
monthly = df.groupby(df['order_date'].dt.to_period('M')).agg(total_revenue=('total_amount', 'sum'), total_orders=('order_id', 'count'))
monthly.plot(kind='line', figsize=(10, 5))
plt.title('Monthly Revenue and Orders')
plt.show()
```

```python
repeat_customers = orders.groupby('customer_id').size().reset_index(name='order_count')
repeat_customers['repeat_customer'] = repeat_customers['order_count'] > 1
print('Repeat purchase rate:', repeat_customers['repeat_customer'].mean())
customer_revenue = orders.groupby('customer_id')['total_amount'].sum().reset_index(name='customer_ltv')
print(customer_revenue.sort_values('customer_ltv', ascending=False).head(10))
```

```python
customers['signup_date'] = pd.to_datetime(customers['signup_date'])
customers['cohort_month'] = customers['signup_date'].dt.to_period('M')
cohort = customers[['customer_id','cohort_month']].merge(orders[['customer_id','order_date']], on='customer_id', how='left')
cohort['order_month'] = pd.to_datetime(cohort['order_date']).dt.to_period('M')
print(cohort.head())
```
